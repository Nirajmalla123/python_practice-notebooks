1. WAP to check whether a mobile number is valid .
  - condition:
    - Exactly 10 digits

2. Wap to check whether a URL is valid.
    - condition:
        - Contains at least one.
        - Starts with WWW

3. WAP to check whether an email is valid.
    - condition:
        - contains exaactly one '@'
        - containing at least one'.'

```python

mobile = input("Enter a mobile number:")
digit = 0

for a in mobile:
    if a.isdigit():
        digit +=1
if len(mobile) == 10 and digit ==10:
    print("valid")
else:
    print("invalid")
```

```python
url = input("Enter URL:")
dot = 0 
for a in url:
    if a == ".":
        dot +=1
if url[:3] == "www" and dot >=1:
    print("valid url")
else:
    print("Invalid URL")
```

```python
email = input("Enter URL:")
dot = 0 
at =0
for a in email:
    if a == ".":
        dot +=1
    elif a=="@":
        at +=1
if at == 1 and dot >=1:
    print("valid url")
else:
    print("Invalid URL")
```

# Control Statement

```python
for i in range(10):
    if i == 5:
        break
    print(i)
```

```python
for i in range(10):
    if i == 5:
        continue                  # it skip the content or number
    print(i)
```
