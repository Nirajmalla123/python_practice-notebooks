wap to find the sum of the primary diagonal elements of a square matrix.

```python
a = [[1,2,3],
     [3,4,5],
     [5,6,7]]
s = 0
for i in range(len(a)):
    s += a[i][i]
print(f"Sum of diagnoal martix = {s}")
```

#### List compression

```python
a = []
for i in range (1,10):
    if i %2 ==0:
        a.append(i)
print(f"Even Number: {a}")
```

```python
a = [i for i in range (1,10) if i %2 ==0]
print(f"Even Number: {a}")
```
