##### Write a Python program (WAP) to search for a specific record (by Name, Age, or Address) inside a nested list of student details. Display all the matching records formatting them clearly, and print the total count of matches found.

```python
def search_student_records(records, query):
    # Convert query to a string and lowercase it for flexible matching
    query_str = str(query).strip().lower()
    matched_list = []

    # Loop through each individual record in the nested list
    for student in records:
        name = student[0]
        age = student[1]
        address = student[2]
        
        # Check if query matches a name fragment, exact age, or address fragment
        if (query_str in name.lower() or 
            query_str == str(age) or 
            query_str in address.lower()):
            matched_list.append(student)

    # Display the results to the user
    print(f"\n--- Search Results for '{query}' ---")
    if matched_list:
        for row in matched_list:
            print(f"Name: {row[0]:<15} | Age: {row[1]:<3} | Address: {row[2]}")
    else:
        print("No records found matching your search.")
        
    print(f"Total matching records: {len(matched_list)}")
    print("-" * 40)


# Pre-defined database stored as a nested list: [Name, Age, Address]
students_db = [
    ["Rohan Sharma", 20, "Kathmandu"],
    ["Sita Rai", 22, "Bhaktapur"],
    ["Amit Patel", 20, "Lalitpur"],
    ["Deepa Shrestha", 23, "Bhaktapur"],
    ["Hari Thapa", 25, "Pokhara"]
]

# Prompt the user for input and run the search function
search_term = input("Enter Name, Age, or Address to search: ")
search_student_records(students_db, search_term)
```
