```python
n = int(input("Enter a number:"))
temp = n 
a = len(str(n))
sumdigit = 0 
for i in range(len(str(n))):
    digit = n % 10 
    fact = 1 
    for j in range(a,digit+1):
        fact +=j
    sumdigit += fact 
    n //=10
if sumdigit == temp:
    print("Strong number")
else:
    print("Not Strong number")
```

```python
n = int(input("Enter a number: "))
temp = n 
a = len(str(n))
sumdigit = 0
for i in range(a):
    digit = n % 10
    sumdigit = sumdigit +digit **a
    n //=10
if sumdigit == temp:
    print("Armstrong number")
else:
    print("Not Armstrong number")
```

```python
# using nested loop 
for i in range(1,6):   # row 
    for j in range (1, i+1): # columns character "x"
        print("x", end=" ")
    print()
    
```

```python
# using nested loop 
for i in range(5, 0, -1):
    for j in range(1, i + 1):
        print("x", end=" ")
    print()
```

```python
# Outer loop counts up, math reverses the rows
for i in range(1, 6):
    for j in range(1, 6 - i + 1):
        print("x", end=" ")
    print()
```
